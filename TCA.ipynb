```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\HP\Downloads\Customer Churn.csv")
df.head()
```

```python
df.info()
```

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ","0")
df["TotalCharges"] = df["TotalCharges"].astype("float")
```

```python
df.info()
```

```python
df.isnull()
```

```python
df.isnull().sum().sum()
```

```python
#converetd 0 and 1 values of senior citizen to yes/no to make it easier to understand
```

```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen'] = df["SeniorCitizen"].apply(conv)
```

```python
import matplotlib .pyplot as plt
import seaborn as sns

sns.countplot(x = 'Churn', data = df)
plt.title("Count of customers by Churn")            
plt.show()
```

```python
gb = df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'], labels = gb.index, autopct = "%1.2f%%")
plt.title("percentage of churned customers", fontsize = 10)
plt.show()
```

```python
#from the given pie chart we can conclufe that 26,54% of our customers has churned out
#not let's explore the reason behind it
```

```python
sns.countplot(x = "gender", data = df, hue = "Churn")
plt.title("gender distribution")
plt.show()
```

```python
plt.figure(figsize = (3,3))
sns.countplot(x = "SeniorCitizen", data = df, hue = "Churn")
plt.title("churn by SeniorCitizen")
plt.show()
```

```python
# Calculate percentage
churn_percent = pd.crosstab(
    df['SeniorCitizen'],
    df['Churn'],
    normalize='index'
) * 100

# Create stacked bar chart
ax = churn_percent.plot(
    kind='bar',
    stacked=True,
    figsize=(7,5),
    colormap='Set2'
)

# Add percentage labels
for container in ax.containers:
    labels = [f'{bar.get_height():.1f}%' if bar.get_height() > 0 else ''
              for bar in container]
    ax.bar_label(container, labels=labels, label_type='center')

plt.title('Churn by Senior Citizen (%)')
plt.xlabel('Senior Citizen (0 = No, 1 = Yes)')
plt.ylabel('% of Total Customers')
plt.legend(title='Churn')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import math

# Select categorical columns
cols = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents',
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies',
    'Contract', 'PaperlessBilling', 'PaymentMethod'
]

# Create subplot layout
rows = 4
cols_count = 4

fig, axes = plt.subplots(rows, cols_count, figsize=(20, 16))

axes = axes.flatten()

for i, col in enumerate(cols):

    # Calculate percentage
    churn_percent = pd.crosstab(
        df[col],
        df['Churn'],
        normalize='index'
    ) * 100

    # Plot stacked bar
    churn_percent.plot(
        kind='bar',
        stacked=True,
        ax=axes[i],
        colormap='Set2',
        legend=False
    )

    # Add percentage labels
    for container in axes[i].containers:
        labels = [
            f'{bar.get_height():.0f}%'
            if bar.get_height() > 0 else ''
            for bar in container
        ]

        axes[i].bar_label(
            container,
            labels=labels,
            label_type='center',
            fontsize=8
        )

    axes[i].set_title(f'{col} vs Churn', fontsize=11)
    axes[i].set_ylabel('% to Total')
    axes[i].set_xlabel('')
    axes[i].tick_params(axis='x', rotation=45)

# Remove unused subplots
for j in range(len(cols), len(axes)):
    fig.delaxes(axes[j])

# Common legend
handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles, labels,
    title='Churn',
    loc='upper center',
    ncol=2
)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
